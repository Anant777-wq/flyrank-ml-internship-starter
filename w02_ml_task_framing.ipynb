## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task Type**: ***Ranking / Scoring***

**Why**: Our goal is not just to say "yes" or "no" to a page needing an update. We need to score and order the pages so an SEO editor knows exactly which ones to prioritize first.





## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target: Observed traffic decline (e.g., a significant drop in 90-day impressions).

Why: We are strictly following the rule that a target must be observed, not defined by human opinion. We are using historical data to predict if a page's traffic is actively decaying.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

***Metric***: Precision@K (specifically, Precision@50).

***Why***: An SEO editor only has the bandwidth to update a limited number of pages a week (e.g., 50). We don't care about the model's overall accuracy across millions of URLs; we only care that the top 50 pages it recommends are highly accurate hits.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML: A simple if-statement (like "IF impressions < 100 THEN refresh") is too rigid and only looks at one or two metrics. Content decay involves tangled, shifting signals (e.g., high search volume + dropping CTR + slipping rank + competition level). ML can weigh dozens of these signals simultaneously to find complex patterns that a human-written rule would miss.
